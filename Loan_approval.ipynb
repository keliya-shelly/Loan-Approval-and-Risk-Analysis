```python
# 4. Check loan approval consistency
df.groupby('Loan_Approved')[['Loan_Amount', 'Approved_Amount']].agg(
    ['count', 'mean', 'min', 'max']
)
```

```python
# 5. Find suspicious approval records
df[df['Loan_Approved'] == 'No']['Approved_Amount'].describe()
```

```python
# Feature Engineering --> more useful for credit-risk analysis.
# 1. Loan-to-Income Ratio
df['Loan_to_Income_Ratio'] = (
    df['Loan_Amount'] / df['Annual_Income']
)
```

```python
# 2. Net Worth
df['Net_Worth'] = (
    df['Total_Assets'] - df['Total_Liabilities']
)
```

```python
# 3. Liquid Assets
df['Liquid_Assets'] = (
    df['Savings_Balance'] + df['Checking_Balance']
)
```

```python
# 4. Approval Gap
df['Approval_Gap'] = (
    df['Loan_Amount'] - df['Approved_Amount']
)
```

```python
# -> Credit Score Band
df['Credit_Score_Band'] = pd.cut(
    df['Credit_Score'],
    bins=[0, 579, 669, 739, 799, 850],
    labels=['Poor', 'Fair', 'Good', 'Very Good', 'Excellent']
)
```

```python
df[['Loan_to_Income_Ratio',
    'Net_Worth',
    'Liquid_Assets',
    'Approval_Gap',
    'Credit_Score_Band']].head()
```

```python
df['Credit_Score_Band'].value_counts().sort_index()
```

```python
# Your First Real Business Analysis
approval_rate = (
    df.groupby('Credit_Score_Band')['Loan_Approved']
      .value_counts(normalize=True)
      .unstack()
      .fillna(0)
      .mul(100)
      .round(2)
)

approval_rate
```

```python
df['Income_Band'] = pd.qcut(
    df['Annual_Income'],
    q=5,
    labels=['Very Low', 'Low', 'Medium', 'High', 'Very High']
)
```

```python
income_approval = (
    df.groupby('Income_Band', observed=True)['Loan_Approved']
      .value_counts(normalize=True)
      .unstack()
      .fillna(0)
      .mul(100)
      .round(2)
)

income_approval
```

```python
# Default Risk vs Credit Score
risk_by_credit = (
    df.groupby('Credit_Score_Band')['Default_Risk']
      .value_counts(normalize=True)
      .unstack()
      .fillna(0)
      .mul(100)
      .round(2)
)

risk_by_credit
```

```python
 # DTI vs Default Risk
df['DTI_Band'] = pd.cut(
    df['Debt_To_Income_Ratio'],
    bins=[0, 10, 20, 30, 100],
    labels=['Low', 'Moderate', 'High', 'Very High']
)
```

```python
dti_risk = (
    df.groupby('DTI_Band', observed=True)['Default_Risk']
      .value_counts(normalize=True)
      .unstack()
      .fillna(0)
      .mul(100)
      .round(2)
)

dti_risk
```

```python
# Late Payments → Default Risk
late_payment_risk = (
    df.groupby('Late_Payments_Last_2_Years')['Default_Risk']
      .value_counts(normalize=True)
      .unstack()
      .fillna(0)
      .mul(100)
      .round(2)
)

late_payment_risk
```

```python
# Advanced Risk Segmentation
df['Risk_Flag'] = (
    (df['Credit_Score'] < 580) |
    (df['Debt_To_Income_Ratio'] > 30) |
    (df['Late_Payments_Last_2_Years'] >= 3)
)
```

```python
risk_approval = (
    df.groupby('Risk_Flag')['Loan_Approved']
      .value_counts(normalize=True)
      .unstack()
      .fillna(0)
      .mul(100)
      .round(2)
)

risk_approval
```
